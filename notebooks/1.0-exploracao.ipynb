```python
import pandas as pd

# 1. Carregar o dataset
df = pd.read_csv('data/raw/ransom.csv')

# 2. Encontrar colunas onde o número de valores únicos (nunique) é igual a 1
constant_cols = [col for col in df.columns if df[col].nunique() <= 1]

# 3. Mostrar os resultados
print(f"Colunas com variância zero ({len(constant_cols)}):")
for col in constant_cols:
    val = df[col].unique()[0]
    print(f" - {col} (Valor constante: {val})")
```
