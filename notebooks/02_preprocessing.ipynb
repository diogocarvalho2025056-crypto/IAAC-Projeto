# ⚙️ Pré-Processamento e Engenharia de Features

Neste notebook, apliquei o pipeline de limpeza aos dados em bruto `data/raw/ransom.csv` e exportamos a versão pronta para treino.

```python
import sys
import os
sys.path.append('../src')

import pandas as pd
from features import clean_and_transform_features

# 1. Carregar dados em bruto
df_raw = pd.read_csv('../data/raw/ransom.csv')
print(f"Dataset original: {df_raw.shape[0]} linhas x {df_raw.shape[1]} colunas")

# 2. Executar transformação
df_clean = clean_and_transform_features(df_raw)
print(f"Dataset transformado: {df_clean.shape[0]} linhas x {df_clean.shape[1]} colunas")

# 3. Garantir que o diretório de destino existe
output_dir = '../data/processed'
os.makedirs(output_dir, exist_ok=True)

# 4. Guardar dataset limpo na pasta data/processed/
output_path = os.path.join(output_dir, 'ransom_cleaned.csv')
df_clean.to_csv(output_path, index=False)
print(f"✅ Ficheiro guardado com sucesso em '{output_path}'")
```
