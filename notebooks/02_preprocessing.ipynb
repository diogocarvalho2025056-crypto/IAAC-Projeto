# ⚙️ Pré-Processamento e Engenharia de Features

Aplica `clean_and_transform_features` ao dataset bruto e guarda a versão pronta para modelação.

Principais decisões:
- **Hexadecimal e decimal na mesma coluna** (`0x00008600` e `34304`): convertidos valor a valor para número. Antes, estas 42 colunas eram descartadas em silêncio por não serem numéricas.
- **Listas em texto** (`DllCharacteristics`, `text_Characteristics`, `rdata_Characteristics`) → uma coluna binária por flag.
- **Categorias pequenas** (`PEType`, `MachineType`, `Subsystem`, ...) → one-hot.
- **`md5` não é feature**, mas é mantido para dividir os dados por ficheiro (o dataset tem 21.752 linhas mas só 13.903 ficheiros únicos).

```python
import sys, os, warnings
sys.path.append('../src')
warnings.filterwarnings('ignore')
import pandas as pd
from features import clean_and_transform_features

df_raw = pd.read_csv('../data/raw/ransom.csv')
print(f"Dataset original: {df_raw.shape[0]} linhas x {df_raw.shape[1]} colunas")
print(f"Ficheiros únicos (md5): {df_raw['md5'].nunique()}")
```

```python
df_clean = clean_and_transform_features(df_raw)
print(f"Dataset transformado: {df_clean.shape[0]} linhas x {df_clean.shape[1]} colunas")
n_feat = df_clean.drop(columns=['md5','Class','Category','Family']).shape[1]
print(f"Features numéricas: {n_feat}")
assert df_clean.drop(columns=['md5','Class','Category','Family']).select_dtypes('number').shape[1] == n_feat
```

```python
os.makedirs('../data/processed', exist_ok=True)
df_clean.to_csv('../data/processed/ransom_cleaned.csv', index=False)
print("✅ Guardado em ../data/processed/ransom_cleaned.csv")
```
